```python
from datetime import date, timedelta
from calendar import monthrange
from pathlib import Path

import pandas as pd


MONTHS_TO_RUN = [
    (2025, 4),
    (2025, 5),
    (2025, 6),
    (2025, 7),
    (2025, 8),
    (2025, 9),
    (2025, 10),

    (2026, 4),
    (2026, 5),
    (2026, 6),
    (2026, 7),
    (2026, 8),
    (2026, 9),
    (2026, 10),
]

today = date.today()

monthly_rows = []

monthly_dir = Path(
    "outputs/prm_v2/monthly_summary"
)

monthly_dir.mkdir(
    parents=True,
    exist_ok=True,
)


for year, month in MONTHS_TO_RUN:

    month_start = date(year, month, 1)

    month_end = date(
        year,
        month,
        monthrange(year, month)[1],
    )

    print(
        "\n"
        + "=" * 80
    )

    print(
        f"Running {month_start:%Y-%m}"
    )

    try:

        kwargs = {}

        run_s25 = False
        run_s26 = False

        # --------------------------------------------------
        # Fully historical month
        # --------------------------------------------------

        if month_end < today:

            run_s25 = True

            kwargs["s25_start"] = str(month_start)
            kwargs["s25_end"] = str(month_end)

        # --------------------------------------------------
        # Fully future month
        # --------------------------------------------------

        elif month_start > today:

            run_s26 = True

            kwargs["s26_start"] = str(month_start)
            kwargs["s26_end"] = str(month_end)

        # --------------------------------------------------
        # Current month split
        # --------------------------------------------------

        else:

            yesterday = today - timedelta(days=1)

            run_s25 = True
            run_s26 = True

            kwargs["s25_start"] = str(month_start)
            kwargs["s25_end"] = str(yesterday)

            kwargs["s26_start"] = str(today)
            kwargs["s26_end"] = str(month_end)

        result = run_s25_s26_v2(

            run_s25=run_s25,
            run_s26=run_s26,

            vehicle_models=VEHICLE_MODELS,
            config=config,

            run_p100=True,
            run_p90=True,

            run_fleet_report=True,

            fleet_report_policy="if_needed",
            current_fleet_only_first=True,

            output_dir=str(monthly_dir),
            output_xlsx_name=f"{year}_{month:02d}.xlsx",

            max_ev10=5,
            max_ev18=5,

            **kwargs,
        )

        # ==========================================
        # P100
        # ==========================================

        p100 = result["outputs_p100"]

        p100_fleet = p100["fleet_utilisation"]

        monthly_rows.append(
            {
                "month": month_start.strftime("%Y-%m"),
                "mode": "P100",

                "arrival_sla_pct":
                    float(
                        p100["sla_summary"]
                        .iloc[0]["arrival_sla_pct"]
                    ),

                "peak_amb":
                    int(
                        p100_fleet.loc[
                            p100_fleet["vehicle_type"] == "Amb",
                            "required_for_schedule"
                        ].sum()
                    ),

                "peak_mini":
                    int(
                        p100_fleet.loc[
                            p100_fleet["vehicle_type"] == "Mini",
                            "required_for_schedule"
                        ].sum()
                    ),

                "peak_staff":
                    float(
                        p100["staff_summary"][
                            "staff_required"
                        ].max()
                    ),

                "fleet_expansion_required":
                    int(
                        result.get(
                            "p100_base_check",
                            pd.DataFrame(
                                [{"needs_fleet_expansion": 0}]
                            )
                        ).iloc[0]["needs_fleet_expansion"]
                    ),

                "current_fleet_sufficient":
                    int(
                        result.get(
                            "p100_base_check",
                            pd.DataFrame(
                                [{"current_fleet_sufficient": 1}]
                            )
                        ).iloc[0]["current_fleet_sufficient"]
                    ),

                "recommended_scenario":
                    (
                        result["recommended_p100"]
                        .iloc[0]["scenario"]
                    )
                    if (
                        "recommended_p100" in result
                        and len(result["recommended_p100"]) > 0
                        and "scenario" in result["recommended_p100"].columns
                    )
                    else "CURRENT_FLEET_OK",
            }
        )

        # ==========================================
        # P90
        # ==========================================

        p90 = result["outputs_p90"]

        p90_fleet = p90["fleet_utilisation"]

        monthly_rows.append(
            {
                "month": month_start.strftime("%Y-%m"),
                "mode": "P90",

                "arrival_sla_pct":
                    float(
                        p90["sla_summary"]
                        .iloc[0]["arrival_sla_pct"]
                    ),

                "peak_amb":
                    int(
                        p90_fleet.loc[
                            p90_fleet["vehicle_type"] == "Amb",
                            "required_for_schedule"
                        ].sum()
                    ),

                "peak_mini":
                    int(
                        p90_fleet.loc[
                            p90_fleet["vehicle_type"] == "Mini",
                            "required_for_schedule"
                        ].sum()
                    ),

                "peak_staff":
                    float(
                        p90["staff_summary"][
                            "staff_required"
                        ].max()
                    ),

                "fleet_expansion_required":
                    int(
                        result.get(
                            "p90_base_check",
                            pd.DataFrame(
                                [{"needs_fleet_expansion": 0}]
                            )
                        ).iloc[0]["needs_fleet_expansion"]
                    ),

                "current_fleet_sufficient":
                    int(
                        result.get(
                            "p90_base_check",
                            pd.DataFrame(
                                [{"current_fleet_sufficient": 1}]
                            )
                        ).iloc[0]["current_fleet_sufficient"]
                    ),

                "recommended_scenario":
                    (
                        result["recommended_p90"]
                        .iloc[0]["scenario"]
                    )
                    if (
                        "recommended_p90" in result
                        and len(result["recommended_p90"]) > 0
                        and "scenario" in result["recommended_p90"].columns
                    )
                    else "CURRENT_FLEET_OK",
            }
        )

        # --------------------------------------
        # SAVE AFTER EVERY MONTH
        # --------------------------------------

        pd.DataFrame(
            monthly_rows
        ).to_excel(
            monthly_dir /
            "monthly_summary.xlsx",
            index=False,
        )

        print(
            f"Completed {month_start:%Y-%m}"
        )

    except Exception as e:

        print(
            f"FAILED {month_start:%Y-%m}"
        )

        print(e)

        # Save progress even if a month fails

        pd.DataFrame(
            monthly_rows
        ).to_excel(
            monthly_dir /
            "monthly_summary.xlsx",
            index=False,
        )

        continue


monthly_summary = pd.DataFrame(
    monthly_rows
)

monthly_summary
```
